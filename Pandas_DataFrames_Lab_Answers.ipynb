## Question 29

Create a DataFrame named inventory_df from equal-length lists. Add a new column using a partially indexed Series so that values appear only for selected row indexes. Display the result, then delete the added column and display the DataFrame again.

```python
inventory_df = pd.DataFrame({
    "Item": ["Pen", "Notebook", "Pencil", "Eraser"],
    "Quantity": [50, 30, 40, 25]
})
stock_series = pd.Series(["Shelf A", "Shelf C"], index=[0, 2])
inventory_df["Shelf"] = stock_series
print("After adding partial Series:")
display(inventory_df)

inventory_df = inventory_df.drop(columns=["Shelf"])
print("After deleting Shelf:")
inventory_df
```

Copy a table with at least four columns from any webpage or spreadsheet and read it into lab_df.
