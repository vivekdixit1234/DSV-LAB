## Question 28

Trace the contents of a, b and c; explain view vs copy.

```python
a = np.array([5, 10, 15, 20, 25, 30])
b = a[1:5]          # view into a
c = a[1:5].copy()   # independent copy
b[1] = 99
c[2] = 77
print('a:', a)
print('b:', b)
print('c:', c)
# Changing b changes a because b is a view. Changing c does not change a because c is a copy.
```

## Question 36

Build a 5 × 5 matrix 21–45; extract central 3 × 3, transpose, multiply, row sums, maximum and tied positions.

```python
a = np.arange(21, 46).reshape(5, 5)
central = a[1:4, 1:4]
transposed = central.T
product = central @ transposed
row_sums = product.sum(axis=1)
max_sum = row_sums.max()
positions = np.where(row_sums == max_sum)[0]
print('Original shape:', a.shape)
print(a)
print('Central shape:', central.shape, '\n', central)
print('Transpose shape:', transposed.shape, '\n', transposed)
print('Product shape:', product.shape, '\n', product)
print('Row sums:', row_sums)
print('Maximum row sum:', max_sum)
print('Zero-based row positions tied at maximum:', positions.tolist())
```

## Question 37

Create a 6 × 4 sales quantity matrix; calculate revenues, top day/product, reduce qualifying prices by 10%, and recalculate.

```python
quantities = np.array([[10, 12, 8, 5],
                       [15, 9, 12, 7],
                       [20, 18, 10, 9],
                       [17, 16, 14, 8],
                       [22, 20, 11, 10],
                       [19, 17, 13, 12]])
original_prices = np.array([150, 240, 90, 320])
daily_revenue = quantities @ original_prices
product_revenue = quantities.sum(axis=0) * original_prices
total_quantity = quantities.sum(axis=0)
print('Daily revenue:', daily_revenue)
print('Revenue per product:', product_revenue)
print('Highest-revenue day:', int(np.argmax(daily_revenue) + 1))
print('Best-selling product number by quantity:', int(np.argmax(total_quantity) + 1))
new_prices = np.where(total_quantity > 100, original_prices * 0.9, original_prices)
print('Original prices:', original_prices)
print('Adjusted prices:', new_prices)
print('Revised total revenue:', (quantities @ new_prices).sum())
```

## Question 38

Generate 100 normal observations with fixed seed; count outliers and clamp to 30–70; compare statistics.

```python
rng = np.random.default_rng(42)
original = rng.normal(50, 12, 100)
outside = (original < 30) | (original > 70)
cleaned = np.where(original < 30, 30, np.where(original > 70, 70, original))
def stats(x):
    return {'mean': x.mean(), 'variance': x.var(), 'std': x.std(), 'min': x.min(), 'max': x.max()}
print('Values outside 30–70:', int(outside.sum()))
print('Original statistics:', stats(original))
print('Cleaned statistics:', stats(cleaned))
# Boundary replacement reduces extremes and constrains the minimum/maximum; mean and spread may change.
```

## Question 39

Combine three branch Series by addition; identify missing values, common quarters, and a separate absent-as-zero result.

```python
branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})
ordinary = branch_a + branch_b + branch_c
print('Ordinary addition (missing values propagate):')
print(ordinary)
common = sorted(set(branch_a.index) & set(branch_b.index) & set(branch_c.index))
print('Quarters common to all branches:', common)
print('Combined earnings for common quarters:')
print(ordinary.reindex(common))
zero_filled = branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0)
print('Absent records treated as zero:')
print(zero_filled)
# Ordinary addition gives NaN if any branch lacks a quarter; fill_value=0 treats absent records as zero.
```

## Question 40

Preserve required index order, identify missing student records, filter valid scores >=70, convert to dict and verify.

```python
scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48}
required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']
s = pd.Series(scores, index=required)
print('Series in required order:')
print(s)
print('Missing student records:', s[s.isnull()].index.tolist())
valid = s[s.notnull() & (s >= 70)]
print('Valid scores >=70:')
print(valid)
filtered_dict = valid.to_dict()
reconstructed = pd.Series(filtered_dict)
print('Dictionary:', filtered_dict)
print('Reconstructed Series:')
print(reconstructed)
print('Labelled values match:', valid.to_dict() == reconstructed.to_dict())
# Missing records should remain missing rather than automatically being treated as zero marks.
```
